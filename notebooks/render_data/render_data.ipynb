```python



permissions = []
role_permissions = []

admin_role_id = roles[0]["id"]
user_role_id = roles[1]["id"]

# Khởi tạo Permission và phân quyền cho Role
for p_def in permission_definitions:
    p_id = str(uuid6.uuid7())
    permissions.append({
        "id": p_id,
        "name": p_def["name"],
        "description": p_def["description"]
    })

    # ADMIN được gán tất cả các quyền
    role_permissions.append({
        "role_id": admin_role_id,
        "permission_id": p_id
    })


# USER chỉ được gán các quyền cơ bản (không có quyền Admin)
    user_permissions = [
        "USER_CREATE_POST", "USER_EDIT_OWN_POST", "USER_DELETE_OWN_POST", "REACT_POST",
        "USER_COMMENT", "USER_DELETE_COMMENT", "USER_UPDATE_FROFILE",
        "BOOKMARK_POST", "REPORT_POST"
    ]

    if p_def["name"] in user_permissions:
        role_permissions.append({
            "role_id": user_role_id,
            "permission_id": p_id
        })

for i in range(TOTAL_USERS):
    user_id = str(uuid6.uuid7())

    email = f"user{i+1}@history.vn"
    status = "ACTIVE"
    created_at = rand_time()

    users.append({
        "id": user_id,
        "created_at": created_at,
        "updated_at": created_at,
        "email": email,
        "password": HASHED_PASSWORD,
        "status": status
    })

    username = f"user{i+1}"
    display_name = fake.name()

    # Lấy avatar theo thứ tự, hết danh sách thì quay lại đầu
    avatar = avatar_urls[i % len(avatar_urls)]

    profiles.append({
        "user_id": user_id,
        "created_at": created_at,
        "updated_at": created_at,
        "avatar_url": avatar,
        "bio": fake.text(max_nb_chars=200),
        "display_name": display_name,
        "username": username
    })

    role_id = roles[0]["id"] if i < 2 else roles[1]["id"]

    user_roles.append({
        "user_id": user_id,
        "role_id": role_id
    })
```

```python
# Hàm bốc ngẫu nhiên thời gian từ 1/1 đến 28/2/2025
def rand_time_jan_feb() -> datetime:
    start_date = datetime(2025, 1, 1, 0, 0, 0)
    # 58 ngày bao phủ toàn bộ tháng 1 và tháng 2
    random_days = random.randint(0, 58)
    random_hours = random.randint(0, 23)
    random_minutes = random.randint(0, 59)
    random_seconds = random.randint(0, 59)
    return start_date + timedelta(
        days=random_days, hours=random_hours, minutes=random_minutes, seconds=random_seconds
    )

# Hàm convert datetime thành chuỗi chuẩn SQL trước khi esc
def ts(dt) -> str:
    if dt is None:
        return "NULL"
    if isinstance(dt, str):
        return dt
    return dt.strftime("%Y-%m-%d %H:%M:%S.%f")[:-3]

# Hàm escape chuỗi và bọc nháy đơn của bạn
def esc(val):
    if val is None:
        return "NULL"
    return "'" + str(val).replace("'", "''") + "'"


# Mốc thời gian hệ thống cố định để thay cho NOW(), tránh bị lệch năm 2026
INIT_TIME_SQL = "'2025-01-01 08:00:00.000'"

sql_lines = []

for r in roles:
    sql_lines.append(f"""
INSERT INTO roles (id, created_at, updated_at, name, description)
VALUES (
{esc(r['id'])},
{INIT_TIME_SQL},
{INIT_TIME_SQL},
{esc(r['name'])},
{esc(r['description'])}
);
""")

# --- Render AppPermissions ---
for p in permissions:
    sql_lines.append(f"""
INSERT INTO app_permissions (id, created_at, updated_at, name, description)
VALUES ({esc(p['id'])}, {INIT_TIME_SQL}, {INIT_TIME_SQL}, {esc(p['name'])}, {esc(p['description'])});
""")

# --- Render Role_Permissions ---
for rp in role_permissions:
    sql_lines.append(f"""
INSERT INTO role_permissions (role_id, permission_id)
VALUES ({esc(rp['role_id'])}, {esc(rp['permission_id'])});
""")

# --- Render Users ---
for u in users:
    sql_lines.append(f"""
INSERT INTO users (id, created_at, updated_at, email, password, status)
VALUES (
{esc(u['id'])},
{esc(ts(u['created_at']))},
{esc(ts(u['updated_at']))},
{esc(u['email'])},
{esc(u['password'])},
{esc(u['status'])}
);
""")

# --- Render Profiles ---
for p in profiles:
    sql_lines.append(f"""
INSERT INTO profiles (
user_id, created_at, updated_at,
avatar_url, bio, display_name, username
)
VALUES (
{esc(p['user_id'])},
{esc(ts(p['created_at']))},
{esc(ts(p['updated_at']))},
{esc(p['avatar_url'])},
{esc(p['bio'])},
{esc(p['display_name'])},
{esc(p['username'])}
);
""")

# --- Render User_Roles ---
for ur in user_roles:
    sql_lines.append(f"""
INSERT INTO user_roles (user_id, role_id)
VALUES (
{esc(ur['user_id'])},
{esc(ur['role_id'])}
);
""")
```

```python
with open("seed_users.sql", "w", encoding="utf-8") as f:
    f.write("\n".join(sql_lines))
```

```python
import os
import re
import json
import time
import random
from pathlib import Path
from datetime import datetime, timedelta
from typing import List, Optional, Dict, Any, Literal

import uuid6
from pydantic import BaseModel, Field
from google import genai
from google.genai import types



def _load_key(colab_secret: str, env_var: str) -> str:
    try:
        from google.colab import userdata as _ud
        val = _ud.get(colab_secret)
        if val:
            print(f"[KEY] '{colab_secret}' loaded from Colab Secrets.")
            return val
    except Exception:
        pass
    val = os.environ.get(env_var, "")
    if val:
        print(f"[KEY] '{env_var}' loaded from environment variable.")
    return val

GEMINI_API_KEY = _load_key("GEMINI_API_KEY", "GEMINI_API_KEY")

MODEL_NAME = "gemini-2.5-flash"
client = genai.Client(api_key=GEMINI_API_KEY)


HISTORY_JSON_PATH = "full_test_VTT.json"
# HISTORY_JSON_PATH = "full_VTT_data.json"
IMAGE_JSON_PATH = "image.json"
USERS_JSON_PATH = "users.json"

OUTPUT_SQL_PATH = "posts_ecosystem.sql"

# TARGET_POST_COUNT = 1000
TARGET_POST_COUNT = 10

SEGMENTS_PER_CYCLE = 2
POSTS_PER_SEGMENT = 2

RANDOM_SEED = 42
random.seed(RANDOM_SEED)

# Nếu Gemini lỗi, script sẽ retry
MAX_RETRIES = 3
RETRY_SLEEP_SECONDS = 3

# Khoảng thời gian sinh post
DEFAULT_USER_CREATED_AT = datetime(2025, 1, 1, 8, 0, 0)
MIN_POST_DATE = datetime(2025, 3, 1, 8, 0, 0)
MAX_POST_DATE = datetime(2026, 5, 20, 22, 0, 0)

# Giới hạn độ dài đoạn gửi vào Gemini
MAX_SEGMENT_CHARS = 7000

# Giới hạn độ dài content post
MIN_CONTENT_CHARS = 500
MAX_CONTENT_CHARS = 3500




class GeneratedPost(BaseModel):
    title: str = Field(
        description="Vietnamese post title, realistic, not clickbait, max 180 characters."
    )
    content: str = Field(
        description="Vietnamese post content. Must be realistic for a historical social network."
    )
    tags: List[str] = Field(
        description="List of 2 to 5 Vietnamese topic tags."
    )


class GeneratedPostBatch(BaseModel):
    posts: List[GeneratedPost] = Field(
        description="Exactly 2 generated posts."
    )


#  SQL HELPERS

def new_uuid() -> str:
    return str(uuid6.uuid7())


def esc(val):
    if val is None:
        return "NULL"
    if isinstance(val, bool):
        return "TRUE" if val else "FALSE"
    if isinstance(val, (int, float)):
        return str(val)
    return "'" + str(val).replace("'", "''") + "'"


def ts(dt: datetime) -> str:
    return dt.strftime("%Y-%m-%d %H:%M:%S.%f")[:-3]


def normalize_tag_name(name: str) -> str:
    name = str(name).strip()
    name = re.sub(r"\s+", " ", name)
    return name[:100]


def clean_text(text: str) -> str:
    text = str(text)
    text = re.sub(r"\r\n", "\n", text)
    text = re.sub(r"\n{3,}", "\n\n", text)
    text = re.sub(r"[ \t]+", " ", text)
    return text.strip()

# cắt chuỗi  nếu vượt quá max_len
def clamp_text(text: str, max_len: int) -> str:
    text = clean_text(text)
    if len(text) <= max_len:
        return text
    return text[:max_len].rstrip()
```

```python

def parse_datetime_safe(value):
    if value is None:
        return DEFAULT_USER_CREATED_AT

    if isinstance(value, datetime):
        return value

    value = str(value).strip()

    patterns = [
        "%Y-%m-%d %H:%M:%S",
        "%Y-%m-%d %H:%M:%S.%f",
        "%Y-%m-%dT%H:%M:%S",
        "%Y-%m-%dT%H:%M:%S.%f",
        "%Y-%m-%d"
    ]

    for p in patterns:
        try:
            return datetime.strptime(value, p)
        except ValueError:
            pass

    return DEFAULT_USER_CREATED_AT


def prepare_existing_users(users):

    if "users" not in globals():
        raise RuntimeError(
            "Không tìm thấy biến users. Bạn cần chạy cell sinh users trước rồi mới chạy cell sinh posts."
        )

    if len(users) < 100:
        raise ValueError(f"Cần ít nhất 100 users, hiện tại chỉ có {len(users)} users.")
    # print(f"{len(users)} \n\n")
    prepared_users = []

    for i, u in enumerate(users[:100]):
        prepared_users.append({
            "id": u["id"],
            "created_at": parse_datetime_safe(u.get("created_at")),
            "persona": None
        })

    # 40 người đầu: ACADEMIC
    for i in range(0, 40):
        prepared_users[i]["persona"] = "ACADEMIC"

    # 30 người tiếp theo: STORYTELLER
    for i in range(40, 70):
        prepared_users[i]["persona"] = "STORYTELLER"

    # 20 người tiếp theo: SKEPTIC
    for i in range(70, 90):
        prepared_users[i]["persona"] = "SKEPTIC"

    # 10 người cuối: MISINFORMED
    for i in range(90, 100):
        prepared_users[i]["persona"] = "MISINFORMED"

    return prepared_users


prepared_users = prepare_existing_users(users)
# print(f"{prepared_users} \n\n\n")

users_by_persona = {
    "ACADEMIC": [u for u in prepared_users if u["persona"] == "ACADEMIC"],
    "STORYTELLER": [u for u in prepared_users if u["persona"] == "STORYTELLER"],
    "SKEPTIC": [u for u in prepared_users if u["persona"] == "SKEPTIC"],
    "MISINFORMED": [u for u in prepared_users if u["persona"] == "MISINFORMED"],
}

print("Loaded existing users:", len(prepared_users))
print({k: len(v) for k, v in users_by_persona.items()})

```

```python

#  tải ảnh từ file nếu thiếu biến nào thì bỏ object đó
def load_images(path: str) -> List[Dict[str, str]]:
    with open(path, "r", encoding="utf-8") as f:
        images = json.load(f)

    cleaned = []

    for img in images:
        public_id = img.get("publicId")
        image_url = img.get("imageUrl")

        if not public_id or not image_url:
            continue

        cleaned.append({
            "publicId": str(public_id).strip(),
            "imageUrl": str(image_url).strip()
        })

    if not cleaned:
        raise ValueError("image.json does not contain valid image assets.")

    return cleaned


image_pool = load_images(IMAGE_JSON_PATH)
print("Loaded images:", len(image_pool))



PAGE_MARKER_RE: re.Pattern = re.compile(r"\n\s*\[\[\[PAGE:(\d+)\]\]\]\s*\n")

def load_history_objects(path: str) -> List[Dict[str, Any]]:
    with open(path, "r", encoding="utf-8") as f:
        data = json.load(f)

    if isinstance(data, dict):
        return [data]

    if isinstance(data, list):
        return data

    raise ValueError("History JSON must be either an object or a list of objects.")


history_objects = load_history_objects(HISTORY_JSON_PATH)
print("Loaded history books:", len(history_objects))


# tách trang
def split_book_into_pages(book_obj: Dict[str, Any]) -> List[Dict[str, Any]]:
    book_name = book_obj.get("book_name", "Unknown Book")
    full_text = book_obj.get("full_text", "")
    footnotes_map = book_obj.get("page_footnotes_map", {}) or {}

    parts = PAGE_MARKER_RE.split(full_text)
    print(f"part: {parts} \n\n\n")

    pages = []

    # Nếu không có marker page
    if len(parts) == 1:
        pages.append({
            "book_name": book_name,
            "page": None,
            "text": clean_text(full_text),
            "footnotes": {}
        })
        return pages

    # parts pattern:
    # [text_before_first_marker, page_number, page_text, page_number, page_text, ...]
    i = 1
    while i < len(parts) - 1:
        page_number = parts[i]
        page_text = parts[i + 1]
        # print(f"page_number: {page_number}  \n {page_text}\n\n")

        page_text = clean_text(page_text)
        # print(f"page text {page_text} \n \n \n ")
        if len(page_text) > 100:
            pages.append({
                "book_name": book_name,
                "page": page_number,
                "text": page_text,
                "footnotes": footnotes_map.get(str(page_number), {})
            })
        i += 2
    return pages


all_pages = []

for book in history_objects:
    all_pages.extend(split_book_into_pages(book))
    print(f"all pages: {all_pages} \n\n\n")

if not all_pages:
    raise ValueError("No usable pages extracted from history JSON.")

print("Extracted pages:", len(all_pages))


def build_continuous_segments(
    pages: List[Dict[str, Any]],
) -> List[Dict[str, Any]]:
    segments = []

    i = 0
    while i < len(pages):
        selected_pages = pages[i:i + 1]

        if not selected_pages:
            break

        book_name = selected_pages[0]["book_name"]
        page_ids = [p["page"] for p in selected_pages if p["page"] is not None]

        combined_text = " ".join(p["text"] for p in selected_pages)
        combined_text = clamp_text(combined_text, MAX_SEGMENT_CHARS)

        combined_footnotes = {}
        for p in selected_pages:
            if p["page"] is not None:
                combined_footnotes[str(p["page"])] = p.get("footnotes", {})

        if len(combined_text) > 300:
            segments.append({
                "book_name": book_name,
                "pages": page_ids,
                "text": combined_text,
                "footnotes": combined_footnotes
            })

        i += 1

    return segments


segments = build_continuous_segments(all_pages)
# random.shuffle(segments)

if not segments:
    raise ValueError("No usable continuous segments created.")

print("Prepared continuous segments:", len(segments))
```

```python
def build_post_plans(total: int) -> List[Dict[str, str]]:
    plans = []

    n_knowledge = int(total * 0.70)
    n_quick = int(total * 0.15)
    n_draft = int(total * 0.10)
    n_defective = total - n_knowledge - n_quick - n_draft

    for _ in range(n_knowledge):
        plans.append({
            "kind": "KNOWLEDGE_SHARE",
            "status": "PUBLISHED",
            "persona": random.choice(["ACADEMIC", "STORYTELLER"])
        })

    for _ in range(n_quick):
        plans.append({
            "kind": "QUICK_THOUGHT",
            "status": "PUBLISHED",
            "persona": random.choice(["STORYTELLER", "SKEPTIC", "ACADEMIC"])
        })

    for _ in range(n_draft):
        plans.append({
            "kind": "DRAFT",
            "status": "DRAFT",
            "persona": random.choice(["ACADEMIC", "STORYTELLER", "SKEPTIC"])
        })

    for _ in range(n_defective):
        plans.append({
            "kind": "DEFECTIVE_CONTENT",
            "status": random.choice(["FLAGGED", "REJECTED"]),
            "persona": "MISINFORMED"
        })

    random.shuffle(plans)
    return plans


post_plans = build_post_plans(TARGET_POST_COUNT)

from collections import Counter
print("Post plan distribution:", Counter([p["kind"] for p in post_plans]))
print("Status distribution:", Counter([p["status"] for p in post_plans]))
print("Persona distribution:", Counter([p["persona"] for p in post_plans]))
```

### prompt + call llm

```python


def format_footnotes(footnotes: Dict[str, Any]) -> str:
    if not footnotes:
        return "No footnotes."

    lines = []

    for page, notes in footnotes.items():
        lines.append(f"Page {page}:")
        if isinstance(notes, dict):
            for note_id, note_text in notes.items():
                lines.append(f"[{note_id}] {note_text}")
        else:
            lines.append(str(notes))

    return "\n".join(lines)

def build_prompt_from_segments(cycle_segments, plan_a, plan_b):
    source_blocks = []

    for idx, segment in enumerate(cycle_segments, start=1):
        book_name = segment["book_name"]
        pages = ", ".join(segment.get("pages", [])) if segment.get("pages") else "Unknown"
        source_text = segment["text"]
        footnotes = format_footnotes(segment.get("footnotes", {}))

        source_blocks.append(f"""
SOURCE SEGMENT {idx}
Book: {book_name}
Pages: {pages}

Text:
{source_text}

Footnotes:
{footnotes}
""".strip())

    combined_sources = "\n\n====================\n\n".join(source_blocks)

    return f"""
You are generating realistic Vietnamese posts for a historical social network about the Trần Dynasty of Vietnam, 1226-1400.

You will receive multiple continuous historical source segments.
Use all provided source segments as background context.
Do not simply summarize each segment separately.
Synthesize them into realistic posts.

HISTORICAL SOURCE SEGMENTS:
{combined_sources}

Generate exactly 2 Vietnamese posts.

POST 1 CONFIG:
- kind: {plan_a["kind"]}
- status: {plan_a["status"]}
- author persona: {plan_a["persona"]}

POST 2 CONFIG:
- kind: {plan_b["kind"]}
- status: {plan_b["status"]}
- author persona: {plan_b["persona"]}

Writing rules:
1. Use Vietnamese.
2. Focus on the Trần Dynasty, Vietnamese medieval history, historical interpretation, events, people, institutions, wars, diplomacy, society, culture, or historiography.
3. The posts must feel like realistic user-generated content on a historical social network.
4. Do not mention that the content was generated by AI.
5. Do not mention JSON, schema, prompt, or system instruction.
6. For KNOWLEDGE_SHARE:
   - Write a rich educational post.
   - Include historical context, explanation, and careful wording.
   - Avoid overclaiming when the source is not clear.
7. For QUICK_THOUGHT:
   - Write a shorter reflection, question, or discussion starter.
8. For DRAFT:
   - Make it look unfinished but still coherent.
   - Do not use placeholder text like TODO.
9. For DEFECTIVE_CONTENT:
   - Create a historically mixed-up or unreliable post.
   - The error should be plausible, not absurd.
   - Do not make it hateful, sexual, or graphically violent.
10. Tags:
   - Return 2 to 5 concise Vietnamese tags.
   - Tags should be reusable taxonomy labels.
   - Examples: "Nhà Trần", "Kháng chiến chống Nguyên", "Nhân vật lịch sử", "Sử liệu", "Ngoại giao".
11. Title:
   - Natural Vietnamese.
   - Not clickbait.
12. Content length:
   - QUICK_THOUGHT: 300-900 characters.
   - DRAFT: 300-1200 characters.
   - KNOWLEDGE_SHARE: 900-2600 characters.
   - DEFECTIVE_CONTENT: 400-1200 characters.

Return only data matching the schema.
""".strip()

# call llm
def call_gemini_for_posts(cycle_segments, plan_a, plan_b):
    prompt = build_prompt_from_segments(cycle_segments, plan_a, plan_b)

    for attempt in range(1, MAX_RETRIES + 1):
        try:
            response = client.models.generate_content(
                model=MODEL_NAME,
                contents=prompt,
                config=types.GenerateContentConfig(
                    temperature=0.85,
                    top_p=0.95,
                    response_mime_type="application/json",
                    response_schema=GeneratedPostBatch,
                )
            )

            parsed = response.parsed
            print(f"parsed {parsed} \n\n\n")

            if parsed is None:
                raise ValueError("Gemini returned no parsed response.")

            posts = parsed.posts

            if len(posts) != 2:
                raise ValueError(f"Expected exactly 2 posts, got {len(posts)}")

            return posts

        except Exception as e:
            print(f"[Gemini Error] attempt {attempt}/{MAX_RETRIES}: {e}")
            if attempt < MAX_RETRIES:
                time.sleep(RETRY_SLEEP_SECONDS)
            else:
                raise
```

### Nếu gọi llm bị lỗi thì tạo bài viết có nội dung mặc định

```python
def fallback_post_from_segments(cycle_segments, plan):
    first_segment = cycle_segments[0]
    return fallback_post(first_segment, plan)


def fallback_post(segment: Dict[str, Any], plan: Dict[str, str]) -> GeneratedPost:
    page_text = segment["text"]
    short_source = clamp_text(page_text, 900)

    if plan["kind"] == "DEFECTIVE_CONTENT":
        title = "Một cách hiểu còn gây tranh luận về sự kiện thời Trần"
        content = (
            "Có ý kiến cho rằng một số sự kiện dưới thời Trần thường bị kể lại theo cách quá đơn giản, "
            "khiến người đọc dễ nhầm lẫn giữa các nhân vật, niên đại và bối cảnh chính trị. "
            "Bài viết này cố gắng nêu lại vấn đề, tuy nhiên thông tin vẫn cần được đối chiếu thêm với sử liệu. "
            f"\n\nTrích đoạn tham khảo: {short_source}"
        )
        tags = ["Nhà Trần", "Tranh luận lịch sử", "Sử liệu"]
    elif plan["kind"] == "QUICK_THOUGHT":
        title = "Một chi tiết đáng suy nghĩ về thời Trần"
        content = (
            "Khi đọc lại sử liệu về thời Trần, mình thấy có nhiều chi tiết nhỏ nhưng mở ra khá nhiều câu hỏi. "
            "Không chỉ là chuyện chiến tranh hay triều chính, mà còn là cách người xưa ghi chép, lựa chọn sự kiện "
            "và nhìn nhận vai trò của từng nhân vật."
        )
        tags = ["Nhà Trần", "Thảo luận lịch sử"]
    elif plan["kind"] == "DRAFT":
        title = "Ghi chú chưa hoàn chỉnh về một vấn đề thời Trần"
        content = (
            "Mình đang ghi lại vài ý ban đầu về đoạn sử liệu này. Có thể cần đọc thêm chú thích và so sánh với "
            "những đoạn khác trước khi kết luận. Điểm đáng chú ý là cách sự kiện được đặt trong mạch chính trị "
            "và xã hội của triều Trần."
        )
        tags = ["Nhà Trần", "Ghi chú", "Sử liệu"]
    else:
        title = "Một góc nhìn từ sử liệu về thời Trần"
        content = (
            "Đoạn sử liệu này cho thấy việc nghiên cứu thời Trần không nên chỉ dừng lại ở các sự kiện lớn, "
            "mà cần chú ý đến bối cảnh, nhân vật, cách ghi chép và những chi tiết phụ trợ. "
            "Những thông tin trong chính sử thường ngắn gọn, vì vậy người đọc cần đặt chúng vào quan hệ với "
            "bối cảnh chính trị, quân sự và xã hội đương thời.\n\n"
            f"Trích đoạn tham khảo: {short_source}"
        )
        tags = ["Nhà Trần", "Sử liệu", "Lịch sử Việt Nam"]

    return GeneratedPost(title=title, content=content, tags=tags)
```

### main

```python
posts_records = []
tags_by_name = {}
post_tags_records = []
post_sources_records = []
post_media_records = []

plan_index = 0
segment_index = 0

while len(posts_records) < TARGET_POST_COUNT:
    cycle_segments = []

    for _ in range(SEGMENTS_PER_CYCLE):
        if segment_index >= len(segments):
            segment_index = 0

        cycle_segments.append(segments[segment_index])
        segment_index += 1

    print(f"Số segment gửi cho một lần gọi LLM: {len(cycle_segments)}")

    if len(posts_records) >= TARGET_POST_COUNT:
        break

    if plan_index >= len(post_plans):
        break

    plan_a = post_plans[plan_index]
    plan_b = post_plans[plan_index + 1] if plan_index + 1 < len(post_plans) else post_plans[plan_index]

    try:
        generated_posts = call_gemini_for_posts(cycle_segments, plan_a, plan_b)
        print(f"dữ liệu trả về là {generated_posts} \n\n")
    except Exception as e:
        print("[Fallback used]", e)
        generated_posts = [
            fallback_post_from_segments(cycle_segments, plan_a),
            fallback_post_from_segments(cycle_segments, plan_b)
        ]

    for gen_post, plan in zip(generated_posts, [plan_a, plan_b]):
        if len(posts_records) >= TARGET_POST_COUNT:
            break

        author = choose_author(plan["persona"])
        created_at = random_post_datetime(author["created_at"])
        updated_at = created_at + timedelta(minutes=random.randint(5, 60 * 24 * 30))

        post_id = new_uuid()

        title = clamp_text(gen_post.title, 500)
        content = clamp_text(gen_post.content, MAX_CONTENT_CHARS)

        if len(content) < 100:
            content = content + "\n\n" + fallback_post_from_segments(cycle_segments, plan).content

        status = plan["status"]
        q_score = quality_score_for(plan["kind"], status)
        counts = default_counts_for_status(status, plan["kind"])

        posts_records.append({
            "id": post_id,
            "author_id": author["id"],
            "title": title,
            "content": content,
            "status": status,
            "quality_score": q_score,
            "bookmark_count": counts["bookmark_count"],
            "comment_count": counts["comment_count"],
            "reaction_count": counts["reaction_count"],
            "report_count": counts["report_count"],
            "created_at": created_at,
            "updated_at": updated_at,
            "deleted_at": None
        })

        # Tags
        clean_tags = []

        for raw_tag in gen_post.tags:
            tag_name = normalize_tag_name(raw_tag)

            if not tag_name:
                continue

            clean_tags.append(tag_name)

        clean_tags = list(dict.fromkeys(clean_tags))

        if not clean_tags:
            clean_tags = ["Nhà Trần", "Lịch sử Việt Nam"]

        for tag_name in clean_tags:
            if tag_name not in tags_by_name:
                tags_by_name[tag_name] = {
                    "id": new_uuid(),
                    "name": tag_name,
                    "description": f"Các bài viết liên quan đến {tag_name}.",
                    "usage_count": 0,
                    "created_at": created_at,
                    "updated_at": updated_at
                }

            tags_by_name[tag_name]["usage_count"] += 1

            post_tags_records.append({
                "post_id": post_id,
                "tag_id": tags_by_name[tag_name]["id"],
                "created_at": created_at
            })

        # Source
        # Vì 1 post bây giờ được sinh từ 4 segments,
        # nên nên tạo nhiều source cho cùng 1 post.
        for segment in cycle_segments:
            source_id = new_uuid()
            pages = segment.get("pages", [])
            first_page = pages[0] if pages else None

            source_title = segment["book_name"]

            if first_page:
                source_title = f"{segment['book_name']} - Trang {first_page}"

            post_sources_records.append({
                "id": source_id,
                "post_id": post_id,
                "title": source_title,
                "author_name": "Vũ Văn Quân",
                "published_year": "2019",
                "url": "https://sachdientu.nxbhanoi.com.vn/ebook-free/10256/0/1"
            })

        # Media
        selected_assets = random_media_assets()

        for order, asset in enumerate(selected_assets):
            now = created_at + timedelta(seconds=order)

            post_media_records.append({
                "id": new_uuid(),
                "post_id": post_id,
                "media_type": "IMAGE",
                "resource_type": "IMAGE",
                "public_id": asset["publicId"],
                "media_url": asset["imageUrl"],
                "display_order": order,
                "created_at": now,
                "updated_at": now
            })

        plan_index += 1

        if len(posts_records) % 50 == 0:
            print(f"Generated posts: {len(posts_records)}/{TARGET_POST_COUNT}")

    time.sleep(5)

print("Done generating.")
print("Posts:", len(posts_records))
print("Tags:", len(tags_by_name))
print("Post Tags:", len(post_tags_records))
print("Sources:", len(post_sources_records))
print("Media:", len(post_media_records))
```

# Comments, Reactions, Bookmarks, & Reports

### import and config and helpers
Tỷ lệ tương tác:

* Bài viết bình thường (Normal) sẽ có từ 3–20 reactions và 1–8 bookmarks.

* Bài viết xu hướng (Viral) sẽ có từ 30–120 reactions và 10–45 bookmarks.

* Bài viết "bẩn/xấu" (Bad post) sẽ bị dính từ 2–6 lượt report.

```python
import random
import time
from datetime import datetime, timedelta
from typing import List, Dict, Any, Optional

import uuid6
from pydantic import BaseModel, Field
from google import genai
from google.genai import types


def _load_key(colab_secret: str, env_var: str) -> str:
    try:
        from google.colab import userdata as _ud
        val = _ud.get(colab_secret)
        if val:
            print(f"[KEY] '{colab_secret}' loaded from Colab Secrets.")
            return val
    except Exception:
        pass
    val = os.environ.get(env_var, "")
    if val:
        print(f"[KEY] '{env_var}' loaded from environment variable.")
    return val

GEMINI_API_KEY = _load_key("GEMINI_API_KEY", "GEMINI_API_KEY")

MODEL_NAME = "gemini-2.5-flash"
client = genai.Client(api_key=GEMINI_API_KEY)


OUTPUT_SQL_PATH = "post_interactions.sql"

# TARGET_COMMENT_COUNT = 3000
TARGET_COMMENT_COUNT = 30

MAX_RETRIES = 3
RETRY_SLEEP_SECONDS = 3

RANDOM_SEED = 42
random.seed(RANDOM_SEED)

REACTION_TYPES = [
    "INFORMATIVE",
    "SURPRISED",
    "SAD",
    "LIKE",
    "LOVE",
    "ANGRY"
]

REPORT_REASON_TYPES = [
    "MISINFORMATION",
    "FAKE_HISTORY"
]

REPORT_STATUS = "PENDING"

TARGET_TYPE_POST = "POST"
TARGET_TYPE_COMMENT = "COMMENT"

# Mỗi lần gọi Gemini cho comment thread
COMMENTS_PER_LLM_CALL_MIN = 4
COMMENTS_PER_LLM_CALL_MAX = 8

# Giới hạn nội dung post gửi vào LLM
MAX_POST_CONTENT_CHARS_FOR_PROMPT = 7000

# Report
MIN_REPORTS_PER_BAD_POST = 2
MAX_REPORTS_PER_BAD_POST = 6

# Reaction/bookmark
MIN_REACTIONS_NORMAL = 3
MAX_REACTIONS_NORMAL = 20
MIN_REACTIONS_VIRAL = 30
MAX_REACTIONS_VIRAL = 120

MIN_BOOKMARKS_NORMAL = 1
MAX_BOOKMARKS_NORMAL = 8
MIN_BOOKMARKS_VIRAL = 10
MAX_BOOKMARKS_VIRAL = 45



def new_uuid() -> str:
    return str(uuid6.uuid7())


def esc(val):
    if val is None:
        return "NULL"
    if isinstance(val, bool):
        return "TRUE" if val else "FALSE"
    if isinstance(val, (int, float)):
        return str(val)
    return "'" + str(val).replace("'", "''") + "'"


def ts(dt: datetime) -> str:
    return dt.strftime("%Y-%m-%d %H:%M:%S.%f")[:-3]


def parse_datetime_safe(value) -> datetime:
    if isinstance(value, datetime):
        return value

    if value is None:
        return datetime(2025, 1, 1, 8, 0, 0)

    value = str(value).strip()

    patterns = [
        "%Y-%m-%d %H:%M:%S",
        "%Y-%m-%d %H:%M:%S.%f",
        "%Y-%m-%dT%H:%M:%S",
        "%Y-%m-%dT%H:%M:%S.%f",
        "%Y-%m-%d"
    ]

    for p in patterns:
        try:
            return datetime.strptime(value, p)
        except ValueError:
            pass

    return datetime(2025, 1, 1, 8, 0, 0)


def clamp_text(text: str, max_len: int) -> str:
    text = str(text or "").strip()
    if len(text) <= max_len:
        return text
    return text[:max_len].rstrip()


def random_after_post_time(post_created_at, min_minutes=5, max_days=90):
    post_created_at = parse_datetime_safe(post_created_at)

    min_dt = post_created_at + timedelta(minutes=min_minutes)
    max_dt = post_created_at + timedelta(days=max_days)

    total_seconds = max(1, int((max_dt - min_dt).total_seconds()))
    return min_dt + timedelta(seconds=random.randint(0, total_seconds))


def get_all_user_ids():
    return [u["id"] for u in prepared_users]

# lấy list user theo tính cách
def get_user_ids_by_persona(persona: str):
    return [u["id"] for u in users_by_persona.get(persona, [])]

# lấy 1 user có tính cách
def choose_user_id_by_persona(persona: str):
    users = get_user_ids_by_persona(persona)
    if not users:
        return random.choice(get_all_user_ids())
    return random.choice(users)

# chọn ngẫu nhiên set user
def choose_distinct_users(count: int, exclude_user_ids=None):
    exclude_user_ids = set(exclude_user_ids or [])
    candidates = [u["id"] for u in prepared_users if u["id"] not in exclude_user_ids]

    if count >= len(candidates):
        return candidates

    return random.sample(candidates, count)

```

```python

class GeneratedComment(BaseModel):
    local_id: str = Field(
        description="Local temporary id such as c1, c2, c3."
    )
    parent_local_id: Optional[str] = Field(
        default=None,
        description="Parent local id for nested replies. Null for root comment."
    )
    persona: str = Field(
        description="One of ACADEMIC, STORYTELLER, SKEPTIC, MISINFORMED."
    )
    content: str = Field(
        description="Vietnamese realistic comment content."
    )


class GeneratedCommentThread(BaseModel):
    comments: List[GeneratedComment] = Field(
        description="A list of realistic Vietnamese comments."
    )


class GeneratedReportReason(BaseModel):
    reason_text: str = Field(
        description="Specific Vietnamese report reason explaining historical misinformation."
    )
```

### prompt

```python
def build_comment_prompt(post: Dict[str, Any], target_count: int) -> str:
    title = post.get("title", "")
    content = clamp_text(post.get("content", ""), MAX_POST_CONTENT_CHARS_FOR_PROMPT)
    status = post.get("status", "PUBLISHED")

    return f"""
You are generating realistic Vietnamese comment threads for a historical social network about the Trần Dynasty of Vietnam.

POST STATUS:
{status}

POST TITLE:
{title}

POST CONTENT:
{content}

Generate exactly {target_count} Vietnamese comments.

Requirements:
1. The comments must form a realistic conversation thread.
2. Use local_id values: c1, c2, c3, ...
3. Some comments must be root comments with parent_local_id = null.
4. Some comments should be nested replies using parent_local_id.
5. Include at least one SKEPTIC persona who challenges historical validity, asks for source, chronology, or interpretation.
6. Include at least one ACADEMIC persona who responds with structured counterpoints, source-aware reasoning, or careful historical clarification.
7. STORYTELLER comments may be more casual, reflective, or community-like.
8. MISINFORMED comments may misunderstand details but must not be hateful or extreme.
9. Do not use profanity.
10. Do not mention AI, prompt, JSON, or schema.
11. Keep comments natural, not too long.
12. Each comment should be between 30 and 450 Vietnamese characters.
13. Avoid copying the post content word-for-word.

Return only data matching the schema.
""".strip()


def call_gemini_for_comment_thread(post: Dict[str, Any], target_count: int) -> List[GeneratedComment]:
    prompt = build_comment_prompt(post, target_count)

    for attempt in range(1, MAX_RETRIES + 1):
        try:
            response = client.models.generate_content(
                model=MODEL_NAME,
                contents=prompt,
                config=types.GenerateContentConfig(
                    temperature=0.85,
                    top_p=0.95,
                    response_mime_type="application/json",
                    response_schema=GeneratedCommentThread,
                )
            )

            parsed = response.parsed

            if parsed is None:
                raise ValueError("Gemini returned no parsed comment response.")

            comments = parsed.comments

            if not comments:
                raise ValueError("Gemini returned empty comments.")

            # Cắt đúng target_count nếu model trả dư
            comments = comments[:target_count]

            return comments

        except Exception as e:
            print(f"[Gemini Comment Error] attempt {attempt}/{MAX_RETRIES}: {e}")

            if attempt < MAX_RETRIES:
                time.sleep(RETRY_SLEEP_SECONDS * attempt)
            else:
                raise
```

### Fallback comment nếu Gemini lỗi

```python
def fallback_comments_for_post(post: Dict[str, Any], target_count: int) -> List[GeneratedComment]:
    templates = [
        {
            "persona": "SKEPTIC",
            "content": "Mình thấy chỗ này cần đối chiếu thêm với sử liệu, nhất là phần niên đại và bối cảnh chính trị."
        },
        {
            "persona": "ACADEMIC",
            "content": "Ý này có thể hợp lý nếu đặt trong mạch sự kiện rộng hơn, nhưng nên phân biệt giữa ghi chép chính sử và cách diễn giải hiện đại."
        },
        {
            "persona": "STORYTELLER",
            "content": "Đọc bài này thấy thời Trần có rất nhiều chi tiết thú vị, không chỉ nằm ở các trận đánh lớn."
        },
        {
            "persona": "ACADEMIC",
            "content": "Theo mình, cần xem thêm nguồn gốc đoạn trích và chú thích đi kèm trước khi kết luận chắc chắn."
        },
        {
            "persona": "SKEPTIC",
            "content": "Có ai biết đoạn này nằm ở trang hoặc sách nào không? Mình muốn kiểm tra lại ngữ cảnh."
        },
        {
            "persona": "STORYTELLER",
            "content": "Nếu triển khai thành một chuỗi bài về nhà Trần thì phần này rất đáng để đào sâu thêm."
        }
    ]

    result = []

    for i in range(target_count):
        t = templates[i % len(templates)]
        parent_local_id = None

        if i >= 2 and random.random() < 0.45:
            parent_local_id = f"c{random.randint(1, i)}"

        result.append(
            GeneratedComment(
                local_id=f"c{i + 1}",
                parent_local_id=parent_local_id,
                persona=t["persona"],
                content=t["content"]
            )
        )

    return result
```

```python


def build_comment_targets(posts: List[Dict[str, Any]], target_total: int):
    """
    Top 20% posts nhận nhiều comment hơn.
    Các post còn lại nhận ít hơn.
    """

    post_count = len(posts)
    top_count = max(1, int(post_count * 0.20))

    # Vì chưa có dữ liệu thật về độ viral, ta chọn top theo quality/status/random.
    # PUBLISHED được ưu tiên hơn DRAFT/REJECTED.
    def score_post(p):
        status = p.get("status", "PUBLISHED")
        quality = float(p.get("quality_score", 0.5))

        status_bonus = {
            "PUBLISHED": 1.0,
            "DRAFT": 0.2,
            "FLAGGED": 0.5,
            "REJECTED": 0.1
        }.get(status, 0.5)

        return quality * status_bonus + random.random() * 0.2

    sorted_posts = sorted(posts, key=score_post, reverse=True)
    viral_post_ids = set(p["id"] for p in sorted_posts[:top_count])

    weights = []

    for p in posts:
        if p["id"] in viral_post_ids:
            weights.append(random.uniform(4.0, 8.0))
        else:
            weights.append(random.uniform(0.4, 1.6))

    total_weight = sum(weights)

    raw_targets = [
        max(0, int(target_total * w / total_weight))
        for w in weights
    ]

    # Sửa sai số do int()
    diff = target_total - sum(raw_targets)

    while diff > 0:
        idx = random.randrange(len(raw_targets))
        raw_targets[idx] += 1
        diff -= 1

    while diff < 0:
        idx = random.randrange(len(raw_targets))
        if raw_targets[idx] > 0:
            raw_targets[idx] -= 1
            diff += 1

    return {
        posts[i]["id"]: raw_targets[i]
        for i in range(len(posts))
    }, viral_post_ids


comment_targets, viral_post_ids = build_comment_targets(
    posts_records,
    TARGET_COMMENT_COUNT
)

print("Viral posts:", len(viral_post_ids))
print("Total target comments:", sum(comment_targets.values()))
```

### tạo comment

```python


comments_records = []

def choose_comment_author_id(persona: str, post_author_id: Optional[str] = None):
    persona = persona if persona in users_by_persona else random.choice(
        ["ACADEMIC", "STORYTELLER", "SKEPTIC"]
    )

    candidates = [
        u["id"]
        for u in users_by_persona.get(persona, [])
        if u["id"] != post_author_id
    ]

    if not candidates:
        candidates = [
            u["id"]
            for u in prepared_users
            if u["id"] != post_author_id
        ]

    return random.choice(candidates)


def generate_comments_for_post(post: Dict[str, Any], target_count: int):
    if target_count <= 0:
        return []

    generated_all = []
    remaining = target_count

    while remaining > 0:
        batch_size = min(
            remaining,
            random.randint(COMMENTS_PER_LLM_CALL_MIN, COMMENTS_PER_LLM_CALL_MAX)
        )

        try:
            batch_comments = call_gemini_for_comment_thread(post, batch_size)
        except Exception as e:
            print("[Comment fallback used]", e)
            batch_comments = fallback_comments_for_post(post, batch_size)

        generated_all.extend(batch_comments)
        remaining -= len(batch_comments)

        time.sleep(1)

    return generated_all[:target_count]


for idx, post in enumerate(posts_records, start=1):
    target_count = comment_targets.get(post["id"], 0)

    if target_count <= 0:
        continue

    post_created_at = parse_datetime_safe(post["created_at"])
    post_author_id = post.get("author_id")

    print(
        f"Comments for post {idx}/{len(posts_records)} | "
        f"target={target_count} | title={post.get('title', '')[:60]}"
    )

    generated_comments = generate_comments_for_post(post, target_count)

    local_to_real_id = {}

    for i, gen_comment in enumerate(generated_comments):
        comment_id = new_uuid()
        local_to_real_id[gen_comment.local_id] = comment_id

        parent_id = None

        if gen_comment.parent_local_id:
            parent_id = local_to_real_id.get(gen_comment.parent_local_id)

        author_id = choose_comment_author_id(
            gen_comment.persona,
            post_author_id=post_author_id
        )

        created_at = random_after_post_time(
            post_created_at,
            min_minutes=10 + i,
            max_days=120
        )

        updated_at = created_at + timedelta(minutes=random.randint(0, 1440))

        comments_records.append({
            "id": comment_id,
            "post_id": post["id"],
            "author_id": author_id,
            "parent_id": parent_id,
            "content": clamp_text(gen_comment.content, 3000),
            "created_at": created_at,
            "updated_at": updated_at,
            "deleted_at": None
        })

    if len(comments_records) >= TARGET_COMMENT_COUNT:
        break

print("Generated comments:", len(comments_records))
```

### tạo reaction and bookmark

```python


reactions_records = []
bookmarks_records = []

used_reaction_pairs = set()
used_bookmark_pairs = set()


def weighted_reaction_type(post_status: str):
    if post_status in ["FLAGGED", "REJECTED"]:
        return random.choices(
            REACTION_TYPES,
            weights=[15, 15, 10, 15, 5, 40],
            k=1
        )[0]

    return random.choices(
        REACTION_TYPES,
        weights=[30, 15, 5, 25, 20, 5],
        k=1
    )[0]


def generate_reactions_and_bookmarks():
    for post in posts_records:
        post_id = post["id"]
        post_author_id = post.get("author_id")
        post_status = post.get("status", "PUBLISHED")
        post_created_at = parse_datetime_safe(post["created_at"])

        is_viral = post_id in viral_post_ids

        if is_viral:
            reaction_target = random.randint(MIN_REACTIONS_VIRAL, MAX_REACTIONS_VIRAL)
            bookmark_target = random.randint(MIN_BOOKMARKS_VIRAL, MAX_BOOKMARKS_VIRAL)
        else:
            reaction_target = random.randint(MIN_REACTIONS_NORMAL, MAX_REACTIONS_NORMAL)
            bookmark_target = random.randint(MIN_BOOKMARKS_NORMAL, MAX_BOOKMARKS_NORMAL)

        # Không cho chính author tự reaction/bookmark bài của mình
        reaction_users = choose_distinct_users(
            reaction_target,
            exclude_user_ids=[post_author_id]
        )

        bookmark_users = choose_distinct_users(
            bookmark_target,
            exclude_user_ids=[post_author_id]
        )

        for user_id in reaction_users:
            pair = (post_id, user_id)

            if pair in used_reaction_pairs:
                continue

            used_reaction_pairs.add(pair)

            created_at = random_after_post_time(
                post_created_at,
                min_minutes=3,
                max_days=150
            )

            reactions_records.append({
                "id": new_uuid(),
                "post_id": post_id,
                "user_id": user_id,
                "type": weighted_reaction_type(post_status),
                "created_at": created_at,
                "updated_at": created_at
            })

        for user_id in bookmark_users:
            pair = (post_id, user_id)

            if pair in used_bookmark_pairs:
                continue

            used_bookmark_pairs.add(pair)

            created_at = random_after_post_time(
                post_created_at,
                min_minutes=5,
                max_days=150
            )

            bookmarks_records.append({
                "id": new_uuid(),
                "post_id": post_id,
                "user_id": user_id,
                "created_at": created_at,
                "updated_at": created_at
            })


generate_reactions_and_bookmarks()

print("Generated reactions:", len(reactions_records))
print("Generated bookmarks:", len(bookmarks_records))
```

### tạo report

```python


def is_bad_post(post: Dict[str, Any]) -> bool:
    status = post.get("status")
    kind = post.get("kind")
    persona = post.get("persona")

    if status in ["FLAGGED", "REJECTED"]:
        return True

    if kind == "DEFECTIVE_CONTENT":
        return True

    if persona == "MISINFORMED":
        return True

    return False


bad_posts = [p for p in posts_records if is_bad_post(p)]

print("Bad posts for reports:", len(bad_posts))


def build_report_prompt(post: Dict[str, Any]) -> str:
    title = post.get("title", "")
    content = clamp_text(post.get("content", ""), MAX_POST_CONTENT_CHARS_FOR_PROMPT)
    status = post.get("status", "UNKNOWN")

    return f"""
You are generating a realistic Vietnamese moderation report reason for a historical social network.

The target post may contain historical misinformation about the Trần Dynasty of Vietnam.

POST STATUS:
{status}

POST TITLE:
{title}

POST CONTENT:
{content}

Generate one specific Vietnamese report reason.

Requirements:
1. The reason must be context-aware.
2. Point out a precise historical concern if possible:
   - wrong date
   - wrong dynasty
   - wrong character
   - mixed-up event
   - unsupported claim
   - distorted interpretation
3. Do not be vague.
4. Do not insult the author.
5. Keep it between 80 and 500 Vietnamese characters.
6. Do not mention AI, JSON, schema, or prompt.

Return only data matching the schema.
""".strip()


def call_gemini_for_report_reason(post: Dict[str, Any]) -> str:
    prompt = build_report_prompt(post)

    for attempt in range(1, MAX_RETRIES + 1):
        try:
            response = client.models.generate_content(
                model=MODEL_NAME,
                contents=prompt,
                config=types.GenerateContentConfig(
                    temperature=0.55,
                    top_p=0.9,
                    response_mime_type="application/json",
                    response_schema=GeneratedReportReason,
                )
            )

            parsed = response.parsed

            if parsed is None:
                raise ValueError("Gemini returned no parsed report reason.")

            reason_text = str(parsed.reason_text).strip()

            if not reason_text:
                raise ValueError("Empty report reason.")

            return clamp_text(reason_text, 1500)

        except Exception as e:
            print(f"[Gemini Report Error] attempt {attempt}/{MAX_RETRIES}: {e}")

            if attempt < MAX_RETRIES:
                time.sleep(RETRY_SLEEP_SECONDS * attempt)
            else:
                raise


def fallback_report_reason(post: Dict[str, Any]) -> str:
    return (
        "Bài viết có dấu hiệu trình bày thông tin lịch sử chưa chính xác hoặc chưa được đối chiếu rõ với sử liệu. "
        "Cần kiểm tra lại niên đại, nhân vật và bối cảnh sự kiện trước khi cho phép hiển thị rộng rãi."
    )


reports_records = []
used_report_pairs = set()

def generate_reports_for_bad_posts():
    moderator_candidates = get_user_ids_by_persona("ACADEMIC")

    for idx, post in enumerate(bad_posts, start=1):
        post_id = post["id"]
        post_author_id = post.get("author_id")
        post_created_at = parse_datetime_safe(post["created_at"])

        report_count = random.randint(
            MIN_REPORTS_PER_BAD_POST,
            MAX_REPORTS_PER_BAD_POST
        )

        reporter_ids = choose_distinct_users(
            report_count,
            exclude_user_ids=[post_author_id]
        )

        print(
            f"Reports for bad post {idx}/{len(bad_posts)} | "
            f"reports={len(reporter_ids)} | title={post.get('title', '')[:60]}"
        )

        try:
            base_reason_text = call_gemini_for_report_reason(post)
        except Exception as e:
            print("[Report fallback used]", e)
            base_reason_text = fallback_report_reason(post)

        for reporter_id in reporter_ids:
            pair = (post_id, reporter_id)

            if pair in used_report_pairs:
                continue

            used_report_pairs.add(pair)

            created_at = random_after_post_time(
                post_created_at,
                min_minutes=20,
                max_days=90
            )

            updated_at = created_at

            reports_records.append({
                "id": new_uuid(),
                "target_id": post_id,
                "target_type": TARGET_TYPE_POST,
                "reporter_id": reporter_id,
                "reason_type": random.choice(REPORT_REASON_TYPES),
                "reason_text": base_reason_text,
                "status": REPORT_STATUS,
                "created_at": created_at,
                "updated_at": updated_at,
                "reviewed_by": None,
                "reviewed_at": None
            })

        time.sleep(1)


generate_reports_for_bad_posts()

print("Generated reports:", len(reports_records))
```

### import, config, helpers

```python
import random
from datetime import datetime, timedelta
from collections import Counter, defaultdict

import uuid6


OUTPUT_SQL_PATH = "social_and_sync.sql"

RANDOM_SEED = 42
random.seed(RANDOM_SEED)

MIN_FOLLOWS_PER_USER = 3
MAX_FOLLOWS_PER_USER = 25

NOTIFICATION_DELAY_SECONDS_MIN = 3
NOTIFICATION_DELAY_SECONDS_MAX = 180

FOLLOW_MIN_AFTER_USER_DAYS = 1


def new_uuid() -> str:
    return str(uuid6.uuid7())


def esc(val):
    if val is None:
        return "NULL"
    if isinstance(val, bool):
        return "TRUE" if val else "FALSE"
    if isinstance(val, (int, float)):
        return str(val)
    return "'" + str(val).replace("'", "''") + "'"


def ts(dt: datetime) -> str:
    return dt.strftime("%Y-%m-%d %H:%M:%S.%f")[:-3]


def parse_datetime_safe(value) -> datetime:
    if isinstance(value, datetime):
        return value

    if value is None:
        return datetime(2025, 1, 1, 8, 0, 0)

    value = str(value).strip()

    patterns = [
        "%Y-%m-%d %H:%M:%S",
        "%Y-%m-%d %H:%M:%S.%f",
        "%Y-%m-%dT%H:%M:%S",
        "%Y-%m-%dT%H:%M:%S.%f",
        "%Y-%m-%d"
    ]

    for p in patterns:
        try:
            return datetime.strptime(value, p)
        except ValueError:
            pass

    return datetime(2025, 1, 1, 8, 0, 0)


def random_datetime_between(start_dt: datetime, end_dt: datetime) -> datetime:
    if start_dt >= end_dt:
        return start_dt + timedelta(minutes=random.randint(1, 60))

    total_seconds = int((end_dt - start_dt).total_seconds())
    return start_dt + timedelta(seconds=random.randint(0, total_seconds))


def notification_time_after(event_created_at):
    event_created_at = parse_datetime_safe(event_created_at)

    return event_created_at + timedelta(
        seconds=random.randint(
            NOTIFICATION_DELAY_SECONDS_MIN,
            NOTIFICATION_DELAY_SECONDS_MAX
        )
    )
```

### follow

```python


def get_user_persona(user_id: str) -> str:
    for persona, users in users_by_persona.items():
        if any(u["id"] == user_id for u in users):
            return persona
    return "UNKNOWN"


def attractiveness_score(user_id: str) -> float:
    persona = get_user_persona(user_id)

    score = 1.0

    if persona == "ACADEMIC":
        score += 4.0
    elif persona == "STORYTELLER":
        score += 2.5
    elif persona == "SKEPTIC":
        score += 1.2
    elif persona == "MISINFORMED":
        score += 0.3

    # Người đăng nhiều bài có khả năng được follow nhiều hơn
    score += min(post_count_by_user[user_id], 20) * 0.35

    return score


def weighted_choose_following_candidates(follower_id: str, count: int):
    candidates = [
        uid for uid in user_ids
        if uid != follower_id
    ]

    weights = [
        attractiveness_score(uid)
        for uid in candidates
    ]

    selected = set()

    while len(selected) < min(count, len(candidates)):
        chosen = random.choices(candidates, weights=weights, k=1)[0]
        selected.add(chosen)

    return list(selected)


def generate_follow_created_at(follower_id: str, following_id: str):
    follower_created_at = user_created_at_by_user_id[follower_id]
    following_created_at = user_created_at_by_user_id[following_id]

    start_dt = max(
        follower_created_at,
        following_created_at
    ) + timedelta(days=FOLLOW_MIN_AFTER_USER_DAYS)

    # Để thỏa điều kiện:
    # User Created At < Follow Event < Post Created At
    end_dt = global_first_post_created_at - timedelta(minutes=10)

    return random_datetime_between(start_dt, end_dt)


follows_records = []
used_follow_pairs = set()

for follower_id in user_ids:
    follower_persona = get_user_persona(follower_id)

    if follower_persona == "ACADEMIC":
        follow_target_count = random.randint(8, 20)
    elif follower_persona == "STORYTELLER":
        follow_target_count = random.randint(6, 18)
    elif follower_persona == "SKEPTIC":
        follow_target_count = random.randint(4, 14)
    else:
        follow_target_count = random.randint(3, 10)

    followings = weighted_choose_following_candidates(
        follower_id,
        follow_target_count
    )

    for following_id in followings:
        pair = (follower_id, following_id)

        if pair in used_follow_pairs:
            continue

        if follower_id == following_id:
            continue

        used_follow_pairs.add(pair)

        created_at = generate_follow_created_at(follower_id, following_id)

        follows_records.append({
            "id": new_uuid(),
            "follower_id": follower_id,
            "following_id": following_id,
            "created_at": created_at,
            "updated_at": created_at
        })

print("Generated follows:", len(follows_records))
```

### notification

```python

notifications_records = []


def add_notification(
    recipient_id,
    actor_id,
    reference_id,
    notification_type,
    content,
    event_created_at
):
    if recipient_id is None:
        return

    if actor_id is not None and recipient_id == actor_id:
        return

    created_at = notification_time_after(event_created_at)

    notifications_records.append({
        "id": new_uuid(),
        "recipient_id": recipient_id,
        "actor_id": actor_id,
        "reference_id": reference_id,
        "type": notification_type,
        "content": content,
        "is_read": random.random() < 0.35,
        "created_at": created_at,
        "updated_at": created_at
    })


# ----------------------------
# COMMENT NOTIFICATIONS
# ----------------------------

for c in comments_records:
    post_id = c["post_id"]
    post = posts_by_id.get(post_id)

    if not post:
        continue

    recipient_id = post["author_id"]
    actor_id = c["author_id"]
    reference_id = c["id"]
    event_created_at = c["created_at"]

    content = "đã bình luận về bài viết của bạn."

    add_notification(
        recipient_id=recipient_id,
        actor_id=actor_id,
        reference_id=reference_id,
        notification_type="COMMENT",
        content=content,
        event_created_at=event_created_at
    )


# ----------------------------
# REACTION NOTIFICATIONS
# ----------------------------

for r in reactions_records:
    post_id = r["post_id"]
    post = posts_by_id.get(post_id)

    if not post:
        continue

    recipient_id = post["author_id"]
    actor_id = r["user_id"]
    reference_id = r["id"]
    event_created_at = r["created_at"]

    reaction_type = r.get("type", "LIKE")
    content = f"đã bày tỏ cảm xúc {reaction_type} với bài viết của bạn."

    add_notification(
        recipient_id=recipient_id,
        actor_id=actor_id,
        reference_id=reference_id,
        notification_type="LIKE",
        content=content,
        event_created_at=event_created_at
    )


# ----------------------------
# FOLLOW NOTIFICATIONS
# ----------------------------

for f in follows_records:
    recipient_id = f["following_id"]
    actor_id = f["follower_id"]
    reference_id = f["id"]
    event_created_at = f["created_at"]

    content = "đã bắt đầu theo dõi bạn."

    add_notification(
        recipient_id=recipient_id,
        actor_id=actor_id,
        reference_id=reference_id,
        notification_type="FOLLOW",
        content=content,
        event_created_at=event_created_at
    )

print("Generated notifications:", len(notifications_records))
```

### validation date and fix

```python


def fix_chronology():
    """
    Tự động chỉnh lại timestamp để đảm bảo:
    - User Created At < Follow Event
    - User Created At < Post Created At
    - Post Created At < Comment/Reaction/Bookmark/Report Created At
    - Event Created At < Notification Created At
    """

    # 1. Fix follows
    for f in follows_records:
        follower_created_at = user_created_at_by_user_id[f["follower_id"]]
        following_created_at = user_created_at_by_user_id[f["following_id"]]

        min_follow_time = max(
            follower_created_at,
            following_created_at
        ) + timedelta(days=1)

        follow_created_at = parse_datetime_safe(f["created_at"])

        if follow_created_at <= min_follow_time:
            follow_created_at = min_follow_time + timedelta(
                minutes=random.randint(5, 60 * 24)
            )

        f["created_at"] = follow_created_at
        f["updated_at"] = max(
            parse_datetime_safe(f.get("updated_at")),
            follow_created_at
        )

    # 2. Fix comments
    for c in comments_records:
        post_created_at = post_created_at_by_post_id.get(c["post_id"])
        author_created_at = user_created_at_by_user_id.get(c["author_id"])

        min_comment_time = max(
            post_created_at,
            author_created_at
        ) + timedelta(minutes=5)

        comment_created_at = parse_datetime_safe(c["created_at"])

        if comment_created_at <= min_comment_time:
            comment_created_at = min_comment_time + timedelta(
                minutes=random.randint(1, 60 * 24 * 30)
            )

        c["created_at"] = comment_created_at
        c["updated_at"] = max(
            parse_datetime_safe(c.get("updated_at")),
            comment_created_at
        )

    # 3. Fix reactions
    for r in reactions_records:
        post_created_at = post_created_at_by_post_id.get(r["post_id"])
        user_created_at = user_created_at_by_user_id.get(r["user_id"])

        min_reaction_time = max(
            post_created_at,
            user_created_at
        ) + timedelta(minutes=3)

        reaction_created_at = parse_datetime_safe(r["created_at"])

        if reaction_created_at <= min_reaction_time:
            reaction_created_at = min_reaction_time + timedelta(
                minutes=random.randint(1, 60 * 24 * 30)
            )

        r["created_at"] = reaction_created_at
        r["updated_at"] = max(
            parse_datetime_safe(r.get("updated_at")),
            reaction_created_at
        )

    # 4. Fix bookmarks
    for b in bookmarks_records:
        post_created_at = post_created_at_by_post_id.get(b["post_id"])
        user_created_at = user_created_at_by_user_id.get(b["user_id"])

        min_bookmark_time = max(
            post_created_at,
            user_created_at
        ) + timedelta(minutes=4)

        bookmark_created_at = parse_datetime_safe(b["created_at"])

        if bookmark_created_at <= min_bookmark_time:
            bookmark_created_at = min_bookmark_time + timedelta(
                minutes=random.randint(1, 60 * 24 * 30)
            )

        b["created_at"] = bookmark_created_at
        b["updated_at"] = max(
            parse_datetime_safe(b.get("updated_at")),
            bookmark_created_at
        )

    # 5. Fix reports
    for r in reports_records:
        reporter_created_at = user_created_at_by_user_id.get(r["reporter_id"])
        report_created_at = parse_datetime_safe(r["created_at"])

        min_report_time = reporter_created_at + timedelta(minutes=10)

        if r["target_type"] == "POST":
            target_created_at = post_created_at_by_post_id.get(r["target_id"])
            min_report_time = max(
                min_report_time,
                target_created_at + timedelta(minutes=10)
            )

        elif r["target_type"] == "COMMENT":
            target_comment = comments_by_id.get(r["target_id"])
            if target_comment:
                target_created_at = parse_datetime_safe(target_comment["created_at"])
                min_report_time = max(
                    min_report_time,
                    target_created_at + timedelta(minutes=10)
                )

        if report_created_at <= min_report_time:
            report_created_at = min_report_time + timedelta(
                minutes=random.randint(1, 60 * 24 * 30)
            )

        r["created_at"] = report_created_at
        r["updated_at"] = max(
            parse_datetime_safe(r.get("updated_at")),
            report_created_at
        )

        if r.get("reviewed_at") is not None:
            reviewed_at = parse_datetime_safe(r["reviewed_at"])

            if reviewed_at <= report_created_at:
                reviewed_at = report_created_at + timedelta(
                    minutes=random.randint(30, 60 * 24 * 7)
                )

            r["reviewed_at"] = reviewed_at

    # 6. Fix notifications
    interaction_created_by_reference_id = {}

    for c in comments_records:
        interaction_created_by_reference_id[c["id"]] = parse_datetime_safe(c["created_at"])

    for r in reactions_records:
        interaction_created_by_reference_id[r["id"]] = parse_datetime_safe(r["created_at"])

    for f in follows_records:
        interaction_created_by_reference_id[f["id"]] = parse_datetime_safe(f["created_at"])

    for n in notifications_records:
        notification_created_at = parse_datetime_safe(n["created_at"])

        reference_created_at = interaction_created_by_reference_id.get(n["reference_id"])
        recipient_created_at = user_created_at_by_user_id.get(n["recipient_id"])

        actor_created_at = None
        if n.get("actor_id"):
            actor_created_at = user_created_at_by_user_id.get(n["actor_id"])

        min_notification_time_candidates = []

        if reference_created_at:
            min_notification_time_candidates.append(
                reference_created_at + timedelta(seconds=5)
            )

        if recipient_created_at:
            min_notification_time_candidates.append(
                recipient_created_at + timedelta(seconds=5)
            )

        if actor_created_at:
            min_notification_time_candidates.append(
                actor_created_at + timedelta(seconds=5)
            )

        if min_notification_time_candidates:
            min_notification_time = max(min_notification_time_candidates)
        else:
            min_notification_time = notification_created_at

        if notification_created_at <= min_notification_time:
            notification_created_at = min_notification_time + timedelta(
                seconds=random.randint(5, 300)
            )

        n["created_at"] = notification_created_at
        n["updated_at"] = max(
            parse_datetime_safe(n.get("updated_at")),
            notification_created_at
        )


def validate_chronology_after_fix():
    errors = []

    # Validate follows
    for f in follows_records:
        follower_created_at = user_created_at_by_user_id[f["follower_id"]]
        following_created_at = user_created_at_by_user_id[f["following_id"]]
        follow_created_at = parse_datetime_safe(f["created_at"])

        if follower_created_at >= follow_created_at:
            errors.append(f"Follow before follower created: {f['id']}")

        if following_created_at >= follow_created_at:
            errors.append(f"Follow before following created: {f['id']}")

        if f["follower_id"] == f["following_id"]:
            errors.append(f"Self follow detected: {f['id']}")

    # Validate comments
    for c in comments_records:
        post_created_at = post_created_at_by_post_id.get(c["post_id"])
        author_created_at = user_created_at_by_user_id.get(c["author_id"])
        comment_created_at = parse_datetime_safe(c["created_at"])

        if post_created_at and post_created_at >= comment_created_at:
            errors.append(f"Comment before post: {c['id']}")

        if author_created_at and author_created_at >= comment_created_at:
            errors.append(f"Comment before user created: {c['id']}")

    # Validate reactions
    for r in reactions_records:
        post_created_at = post_created_at_by_post_id.get(r["post_id"])
        user_created_at = user_created_at_by_user_id.get(r["user_id"])
        reaction_created_at = parse_datetime_safe(r["created_at"])

        if post_created_at and post_created_at >= reaction_created_at:
            errors.append(f"Reaction before post: {r['id']}")

        if user_created_at and user_created_at >= reaction_created_at:
            errors.append(f"Reaction before user created: {r['id']}")

    # Validate bookmarks
    for b in bookmarks_records:
        post_created_at = post_created_at_by_post_id.get(b["post_id"])
        user_created_at = user_created_at_by_user_id.get(b["user_id"])
        bookmark_created_at = parse_datetime_safe(b["created_at"])

        if post_created_at and post_created_at >= bookmark_created_at:
            errors.append(f"Bookmark before post: {b['id']}")

        if user_created_at and user_created_at >= bookmark_created_at:
            errors.append(f"Bookmark before user created: {b['id']}")

    # Validate reports
    for r in reports_records:
        reporter_created_at = user_created_at_by_user_id.get(r["reporter_id"])
        report_created_at = parse_datetime_safe(r["created_at"])

        if reporter_created_at and reporter_created_at >= report_created_at:
            errors.append(f"Report before reporter created: {r['id']}")

        if r["target_type"] == "POST":
            post_created_at = post_created_at_by_post_id.get(r["target_id"])

            if post_created_at and post_created_at >= report_created_at:
                errors.append(f"Report before post: {r['id']}")

        elif r["target_type"] == "COMMENT":
            target_comment = comments_by_id.get(r["target_id"])

            if target_comment:
                comment_created_at = parse_datetime_safe(target_comment["created_at"])

                if comment_created_at >= report_created_at:
                    errors.append(f"Report before comment: {r['id']}")

    # Validate notifications
    interaction_created_by_reference_id = {}

    for c in comments_records:
        interaction_created_by_reference_id[c["id"]] = parse_datetime_safe(c["created_at"])

    for r in reactions_records:
        interaction_created_by_reference_id[r["id"]] = parse_datetime_safe(r["created_at"])

    for f in follows_records:
        interaction_created_by_reference_id[f["id"]] = parse_datetime_safe(f["created_at"])

    for n in notifications_records:
        notification_created_at = parse_datetime_safe(n["created_at"])
        reference_created_at = interaction_created_by_reference_id.get(n["reference_id"])
        recipient_created_at = user_created_at_by_user_id.get(n["recipient_id"])

        actor_created_at = None
        if n.get("actor_id"):
            actor_created_at = user_created_at_by_user_id.get(n["actor_id"])

        if reference_created_at and reference_created_at >= notification_created_at:
            errors.append(f"Notification before referenced event: {n['id']}")

        if recipient_created_at and recipient_created_at >= notification_created_at:
            errors.append(f"Notification before recipient created: {n['id']}")

        if actor_created_at and actor_created_at >= notification_created_at:
            errors.append(f"Notification before actor created: {n['id']}")

    if errors:
        print("Chronology still has errors after auto-fix.")
        print("Error count:", len(errors))

        for e in errors[:30]:
            print("-", e)

        raise RuntimeError("Chronology auto-fix failed.")

    print("Chronology auto-fix passed.")




fix_chronology()
validate_chronology_after_fix()
```

```python

def render_case_update(
    table_name: str,
    id_column: str,
    value_column: str,
    values_by_id: Dict[str, int]
):
    if not values_by_id:
        return ""

    ids = list(values_by_id.keys())

    lines = []
    lines.append(f"UPDATE {table_name}")
    lines.append(f"SET {value_column} = CASE {id_column}")

    for entity_id in ids:
        lines.append(f"    WHEN {esc(entity_id)} THEN {int(values_by_id[entity_id])}")

    lines.append(f"    ELSE {value_column}")
    lines.append("END")
    lines.append(f"WHERE {id_column} IN (")
    lines.append("    " + ",\n    ".join(esc(entity_id) for entity_id in ids))
    lines.append(");")

    return "\n".join(lines)


def render_post_counter_update(column_name: str, counter: Counter):
    values = {
        p["id"]: int(counter.get(p["id"], 0))
        for p in posts_records
    }

    return render_case_update(
        table_name="posts",
        id_column="id",
        value_column=column_name,
        values_by_id=values
    )


def render_tag_usage_update():
    values = {
        tag["id"]: int(usage_count_by_tag.get(tag["id"], 0))
        for tag in tags_by_name.values()
    }

    return render_case_update(
        table_name="tags",
        id_column="id",
        value_column="usage_count",
        values_by_id=values
    )


def render_profile_counter_update(column_name: str, counter: Counter):
    values = {
        u["id"]: int(counter.get(u["id"], 0))
        for u in prepared_users
    }

    return render_case_update(
        table_name="profiles",
        id_column="user_id",
        value_column=column_name,
        values_by_id=values
    )

sql_lines = []


# sql_lines.append("BEGIN;\n\n")



for f in follows_records:
    sql_lines.append(f"""
INSERT INTO follows (
    id,
    follower_id,
    following_id,
    created_at,
    updated_at
)
VALUES (
    {esc(f['id'])},
    {esc(f['follower_id'])},
    {esc(f['following_id'])},
    {esc(ts(f['created_at']))},
    {esc(ts(f['updated_at']))}
);
""")



for n in notifications_records:
    sql_lines.append(f"""
INSERT INTO notifications (
    id,
    recipient_id,
    actor_id,
    reference_id,
    type,
    content,
    is_read,
    created_at,
    updated_at
)
VALUES (
    {esc(n['id'])},
    {esc(n['recipient_id'])},
    {esc(n['actor_id'])},
    {esc(n['reference_id'])},
    {esc(n['type'])},
    {esc(n['content'])},
    {esc(n['is_read'])},
    {esc(ts(n['created_at']))},
    {esc(ts(n['updated_at']))}
);
""")



sql_lines.append(render_post_counter_update("comment_count", comment_count_by_post))

sql_lines.append(render_post_counter_update("reaction_count", reaction_count_by_post))

sql_lines.append(render_post_counter_update("bookmark_count", bookmark_count_by_post))

sql_lines.append(render_post_counter_update("report_count", report_count_by_post))



sql_lines.append(render_tag_usage_update())


sql_lines.append(render_profile_counter_update("follower_count", follower_count_by_user))

sql_lines.append(render_profile_counter_update("following_count", following_count_by_user))


# sql_lines.append("\nCOMMIT;\n")

with open(OUTPUT_SQL_PATH, "w", encoding="utf-8") as f:
    f.write("\n".join(sql_lines))

print(f"SQL file written: {OUTPUT_SQL_PATH}")
print("Follows:", len(follows_records))
print("Notifications:", len(notifications_records))
print("Post counter sync included.")
print("Tag usage sync included.")
print("Profile follower/following sync included.")
```
